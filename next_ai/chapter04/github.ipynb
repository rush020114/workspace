```python
# Squash Merge로 커밋 정리하기
# GitHub PR의 Merge 버튼 옆 ▼ 메뉴에는 3가지 머지 옵션이 존재

# Create a merge commit
# 기본값. 모든 커밋 히스토리를 보존하면서 합침. 가장 안전하지만 히스토리가 복잡해질 수 있습니다.
# Squash and merge
# 실무에서 가장 인기 있는 전략. feature 브랜치의 모든 커밋을 1개로 압축. main 히스토리는 깔끔, 작업 과정 디테일은 사라짐.
# Rebase and merge
# 선형 히스토리. 커밋을 main 위로 옮겨 일자로 만듭니다. 깔끔하지만 커밋 해시가 바뀌므로 공유 브랜치엔 주의.
# 📌 어떤 전략을 쓸까?
# 대부분의 실무 팀은 Squash and merge를 기본 전략으로 채택. feature 브랜치의 잔잔한 작업 커밋은 묻고, main 히스토리는 깔끔하게 유지하기 좋기 때문.
```

```python
# 작업하던 내용을 커밋하지 않고 임시로 넣어두는(치워두는) 명령어
# git stash
# → 지금 작업 중인 변경사항(아직 commit 안 한 것들)을 잠깐 서랍에 넣어두고, 작업 디렉토리를 마지막 커밋 상태로 깨끗하게 되돌림
# git stash pop
# → 서랍에 넣어뒀던 변경사항을 다시 꺼내옴
# 언제 쓰나 (혼자 작업할 때 흔한 상황)

# 작업하다가 급하게 다른 브랜치로 옮겨야 할 때

# 예: feature 브랜치에서 코드 짜던 중인데, 갑자기 main에서 확인할 게 생김
# 근데 지금 코드는 완성 안 돼서 commit하긴 애매함
# 이럴 때 git stash로 잠깐 치워두고 브랜치 이동, 볼일 다 보고 돌아와서 git stash pop으로 이어서 작업


# git pull 하려는데 로컬에 변경사항이 있어서 충돌 경고 뜰 때

# 커밋하기엔 애매한 상태인데 pull은 받아야 할 때, stash로 치워두고 pull 받은 다음 다시 꺼냄



# 자주 쓰는 명령어 조합
# bashgit stash              # 임시 저장
# git stash list          # 저장된 stash 목록 보기
# git stash pop           # 가장 최근 것 꺼내면서 삭제
# git stash apply         # 꺼내지만 목록엔 남겨둠
# git stash drop          # 그냥 삭제
```
