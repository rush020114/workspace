```python
# class 선언
class Post:
  # 생성자 만들기
  # self는 인스턴스 자기 자신을 가리키는 키워드
  # self.content는 class의 속성
  def __init__(self, user, content):
    self.user = user
    self.content = content
    self.comments = []
    self.likers = []

  def like(self, user):
    self.likers.append(user)

  def comment(self, user, content):
    self.comments.append(Comment(user, content))

class Comment:
  def __init__(self, user, content):
    self.user = user
    self.content = content
```

```python
# 상속 클래스 선언
class ImagePost(Post):
  def __init__(self, user, content):
    # super(): 부모 클래스에 접근할 때 사용하고 부모의 클래스가 반환된다.
    # 부모의 생성자 실행  
    super().__init__(user, content)
    self.image = []
```

```python
my_post = ImagePost("alice", "강남맛집")
# 부모의 모든 것을 물려받기 때문에 ImagePost에서도 likers를 출력할 수 있다.

my_post.like("bob")
print(my_post.likers)
```

```python
class LinkPost(Post):
  def __init__(self, author, content, url):
    super().__init__(author, content)
    if not url.startswith("https://") and not url.startswith("http://"):
      # raise로 오류를 보여주면서 생성자 생성을 막을 수 있다.
      raise ValueError("주소 형식을 확인해주세요.")
    self.url = url
    
```

```python
# 다형성
# # 주의할 점: 형제 클래스가 있을 때는 부모 클래스를 직접 인스턴스화하지 않는다. (예상치 못한 버그 발생 가능성 높아짐)
class Reaction:
  def __init__(self, type, post, user):
    self.type = type
    self.post = post
    self.user = user

# 형제 클래스로 type을 확인할 수 있다.
class Like(Reaction):
  def __init__(self, post, user):
    super().__init__("LIKE", post, user)

class Sad(Reaction):
  def __init__(self, post, user):
    super().__init__("SAD", post, user)
```

```python
# 오버라이딩
# 부모의 메소드를 같은 이름으로 재정의하여 커스터마이징하기 위함
# 현재 코드의 예시 부모의 comment를 덮어씌워 댓글을 달 수 없게 만듦
# 또는 Confirm 등으로 정말로 댓글을 달 거냐는 alert 같은 것을 띄울 수도 있음
class ProtectedPost(Post):
  def comment(self, user, content):
    print("Can't comment on protected posts.")
```
