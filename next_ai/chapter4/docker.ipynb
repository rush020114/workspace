# docker

```python
# kernel: 하드웨어를 제어하는 영역
# shell: 사람이 직접 kernel에 명령을 내릴 수 없으니까 interface 역할을 담당하는 영역
# terminal: shell을 실행하는 창 (우리가 실제로 보는 화면)

# chroot → 컨테이너가 자기 폴더 밖을 못 보게 막는 것
# namespace  → 컨테이너끼리 서로 안 보이게 격리시켜주는 것
# cgroup     → 컨테이너가 CPU, 메모리 얼마나 쓸지 제한하는 것

# docker의 개념
# 인프라 환경(OS)에 구애받지 않는 애플리케이션 독립 실행 환경

# docker의 구성요소
#
# client: docker 명령어를 입력하는 영역
# server (Docker Engine)
# ├── Docker Host: 컨테이너가 실제로 실행되는 환경
# └── Docker Registry: 이미지를 저장하고 공유하는 저장소

# Docker daemon: 클라이언트의 명령을 받고, 이미지, 컨테이너, 네트워크 등을 직접 관리
# Image: 컨테이너를 실행하기 위한 탬플릿, 컨테이너 실행을 위한 명령어와 환경 설정 포함
# container: 이미지를 실행한 실제 형태

# 자주 쓰는 명령어
# ├── docker build     → image 만들기
# ├── docker run       → container 실행
# ├── docker ps        → 실행 중인 container 목록
# ├── docker stop      → container 멈추기
# ├── docker rm        → container 삭제
# ├── docker images    → image 목록
# ├── docker logs      → 로그 보기
# └── docker exec -it  → container 안으로 들어가기

# 실무에서 진짜 많이 쓰는 건
# └── docker-compose up
# └── docker-compose down

# docker run
# 자주 씀
# ├── -it    → container 안에 들어갈 때
# ├── -d     → 백그라운드 실행 (서버 띄울 때)
# └── -p     → 포트 연결 (외부에서 접근할 때)

# docker build
# 자주 씀
# ├── -t     → image 이름 지정
# └── -f     → Dockerfile 위치 지정

# Dockerfile 명령어(image 만들 때)
# FROM:    어떤 환경에서 시작할지
# WORKDIR: 컨테이너 안에서 작업할 폴더 지정
# COPY:    파일이나 디렉터리를 컨테이너 내부로 복사
# RUN:     image 빌드할 때 실행되는 명령어 (환경 준비)
# EXPOSE:  컨테이너가 사용할 포트 선언
# CMD:     컨테이너 실행될 때 자동으로 실행할 명령어
```
