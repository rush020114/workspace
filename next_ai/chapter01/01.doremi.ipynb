```python
# append: 오직 한 개의 자료를 마지막에 넣을 수 있는 함수
a = []
b = ['a', 'b', 'c']

a.append(10)
b.append('d')
print(a, b)

# insert: 몇 번째 인덱스에 넣고 싶은 자료를 하나만 넣을 수 있는 함수
# 첫 번째 매개변수: 넣을 인덱스, 두 번째 매개변수: 들어갈 값
c = [1, 2, 4, 5]
c.insert(2, 3)
print(c)

# 원소 하나를 지울 수 있는데 중복될 경우 첫 번째에 발견된 원소가 지워진다.
my_list = [3, 1, 2, 3]
my_list.remove(3)
print(my_list)

# 정렬 함수
# 숫자, 문자가 함께 있으면 사용 불가
e = [2, 1, 4, 3]
f = ['d', 'c', 'b']

e.sort()
f.sort()
print(e, f)

# 꺼내기 함수 
my_list = [2, 3, 5, 7, 11]
# 0번째 요소 꺼내기
print(my_list.pop(0))
# 매개변수가 없으면 0번째 요소를 빼냄
print(my_list.pop())
print(my_list)

# 함수 안에 있는 자료가 몇 개인지 알 수 있는 함수
my_seq = [2, 2, 2, 4, 4]
print(my_seq.count(4))
my_str = "elice and alice"
print(my_str.count('e'))

# 문자열을 쪼개 리스트화하는 함수
# 매개변수 미전달 시 공백을 기준으로 처리
my_str = "1 2 3 4 5"
print(my_str.split())
element = "Na, Mg, Al, Si"
print(element.split(','))

# 리스트를 합쳐서 문자열을 만드는 함수
my_list = ['e', 'l', 'i', 'c', 'e']
print("".join(my_list))
station = ['서울', '대전', '대구']
print("->".join(station))
```

```python
# 딕셔너리도 대괄호로 value를 꺼낸다.
person = {
  "name": "Rabbit",
  "age": 22
}

print(person["name"])

# 자료 추가하기
person["mail"] = "rabbit@elice.com"
print(person)

# 자료 삭제하기
del person["age"]
print(person)

# key들만 추출하기
print(person.keys())

# value만 추출하기
print(person.values())

# (key, value) 형태의 리스트로 추출하기
print(person.items())

# key는 중복될 수 없고 변할 수 없는 자료형(튜플 사용 가능)이어야 한다.
dictionary_example = {(1, 2, 3): "hello"}
print(dictionary_example)
```

```python
# list comprehension: 리스트를 간단하고 효율적이게
# 문법 활용1: [표현식 for 변수 in 시퀀스]
# 흐름
# 1. 시퀀스 자료형에서 자료를 하나씩 가져와서 변수에 담아
# 2. 변수를 지정한 표현식으로 변환
# 3. list에 그대로 담김
# 예) 
new_list = [num * 10 for num in range(10)]
# 1. 0 ~ 9까지의 숫자를 하나씩 가져와서 변수에 담아
# 2. 변수를 * 10하여 리스트의 인덱스를 채움
# 3. 각 숫자의 * 10이 리스트에 담김
print(new_list)

# 문법 활용2: [표현식 for 변수 in 시퀀스 if 조건]
# 흐름
# 1. 시퀀스 자료형에서 자료를 하나씩 가져와서 변수에 담아
# 2. 변수가 지정한 조건에 맞는지 확인
# 3. 조건을 통과한 변수를 지정한 표현식으로 변환
# 4. list에 그대로 담김
# 예)
new_list = [str(num) for num in range(10) if num % 2 == 0]
# 1. 0 ~ 9까지의 숫자를 하나씩 가져와서 변수에 담아
# 2. 각각의 숫자를 차례차례 조건 확인
# 2. 조건에 맞는 변수만 찾아 해당 변수 문자열화하여 리스트의 인덱스를 채움
# 3. 각 숫자의 문자열화가 리스트에 담김
print(new_list)

# 문법 활용3: [표현식1 if 조건 else 표현식2 for 변수 in 시퀀스]
# 흐름: 반복되는 동안 조건이 맞으면 표현식1을 실행 틀리면 2를 실행
# 예)
original_price = [800, 1200, 950, 1100]
discounted_prices = [price * 0.9 if price >= 1000 else price for price in original_price]
print(discounted_prices)
# 해당 리스트에서 가격이 1000원 이상이면 10%할인하고 아니면 그대로 가격을 리스트에 담기
```

```python
my_list = ["Spring", "Summer", "Fall", "Winter"]

# enumerate: 순서가 있는 자료형(str, list, tuple)을 대상으로 인덱스와 그 값을 전달
# 형태: iterator - (0, 값1), (1, 값2), (2, 값3), ..., (n, 값n)
print(enumerate(my_list))
print(list(enumerate(my_list)))

temperatures = {
  "서울": 28,
  "부산": 30,
  "대구": 29,
  "광주": 31
}

# 튜플 안에 튜플이 있는 구조라 구조분해할당을 다음과 같이 진행
# (place, (city, temp)), (place, (city, temp)), (place, (city, temp))
for place, (city, temp) in enumerate(temperatures.items(), start=1):
  print(f"{place}. {city}의 온도: {temp}도")
```

```python
# lambda: 익명 함수(이름 없이 사용할 수 있는 한 줄 함수)
# 문법: lambda 매개변수: 표현식
# 표현식은 return이라 보면 된다.

calcualate_total_price = lambda item_price, quantity: item_price * quantity * 1.1
print(calcualate_total_price(100, 52))

# lambda는 map 함수와 활용하기 좋다.
# map함수: iterable의 자료 하나씩 함수를 적용
# 문법: map(함수, iterable)
# 예)
names = ["rabbit", "elice", "clock"]
upper_name = map(str.upper, names)
print(list(upper_name))

# lambda가 함수이므로 map 함수에 집어넣기 좋다.
square = lambda x: x ** 2
num_list = [10, 11, 12, 13, 14, 15]
map_function = map(square, num_list)
print(list(map_function))

list1 = [1, 2, 3, 4]
list2 = [11, 22, 33, 44]

# map은 병렬로 처리도 가능하기 때문에 두 리스트의 같은 인덱스를 병렬로 처리해준다.
sum_result = list(map(lambda x, y: x + y, list1, list2))
print(sum_result)
```
