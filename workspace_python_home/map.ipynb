```python
# map 정의 : 리스트나 튜플 같은 반복 가능한 자료의 각 요소에 함수를 적용하여 새로운 값을 만들어주는 함수
# map 문법 : map(함수, 반복 가능한 자료)

# iterator : 데이터를 하나씩 꺼낼 수 있는 반복자 객체
#            반복문 등으로 바로 사용하고 나면 소멸되기에 list() 같은 자료구조에 저장해두어야 함
# iterable의 데이터는 각각의 데이터를 iter()에 담아야 iterator가 됨.
# map 특징 : map의 반환값은 iterator
```

```python
number = [1, 2, 3, 4, 5]
squared = map(lambda a : a ** a, number)
print(list(squared))
```
