```python
# f = open("파일.txt", "r") 파일 열기
# f.close()                파일 닫기

# r | 읽기 (파일 없으면 오류)
# w | 쓰기 (파일 없으면 생성, 있으면 덮어씀)
# x | 쓰기 (파일 없을 때만 생성, 있으면 오류)
# a | 이어쓰기 (파일 없으면 생성, 있으면 뒤에 추가)

# b | 바이너리 모드 (rb, wb)
# r+ : 읽기+쓰기, 파일 없으면 오류, 기존 내용 유지
# w+ : 읽기+쓰기, 파일 없으면 생성, 기존 내용 전부 삭제

# encoding : 문자 인코딩 지정 (한글 깨질 때 utf-8 또는 euc-kr 사용)
# - utf-8  : 전세계 표준, 대부분의 환경에서 사용
# - euc-kr : 옛날 한국 윈도우 환경에서 만든 파일에 사용
# read()      : 파일 전체를 문자열 하나로 반환
# readline()  : 한 줄만 반환
# readlines() : 모든 줄을 리스트로 반환

# write()      : 문자열 하나를 파일에 씀
# writelines() : 리스트를 파일에 씀 (줄바꿈 자동X, 직접 \n 넣어야 함)
```
