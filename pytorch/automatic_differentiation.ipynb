```python
# 딥러닝의 매개변수 : 학습되는 값들 = 가중치 + 편향
# 역전파 : 신경망 학습에 가장 많이 사용되는 알고리즘
# 이 알고리즘(역전파)에서 매개변수(학습되는 값)는 손실 함수에 대한 변화도에 따라 조정
# 이러한 변화도를 계산하기 위해 pytorch는 자동 미분을 지원
```

```python
import torch

# 학습시킬 대상만 변화도(gradient)를 계산한다. (가중치, 편향)
x = torch.ones(5) # input tensor
y = torch.zeros(3) # expected output
# 역전파 때 변화도를 계산할 수 있게끔 True로 설정
w = torch.randn(5, 3, requires_grad=True)
b = torch.randn(3, requires_grad=True)
z = x @ w + b
# loss계산 : 예측값과 정답의 차이를 계산
loss = torch.nn.functional.binary_cross_entropy_with_logits(z, y) 

print(f"Gradient function for z = {z.grad_fn}")
print(f"Gradient function for loss = {loss.grad_fn}")
```

```python
# 신경망에서 매개변수의 가중치를 최적화하려면 매개변수에 대한 손실함수의 도함수를 계산
# 매개변수를 최적화하기 위해 backward로 gradient를 구한다.(loss를 통해 얼마나 예측이 빗나간지를 측정한 후 이에 기반해 변화율을 계산)
loss.backward()
print(w.grad)
print(b.grad)
```
