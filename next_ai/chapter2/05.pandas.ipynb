```python
import pandas as pd


df = pd.read_csv("park_visitors.csv")

# apply: Series나 DataFrame의 각 요소에 함수를 적용하는 method
# 눈, 비, 눈/비를 눈/비 하나로 통일
df["날씨"] = df["날씨"].apply(lambda e: "눈/비" if e == "눈" or e == "비" else e)

print(df["날씨"].value_counts())

# data 분포
print(df.describe())

# 날씨별로 총계의 평균
print(df.groupby("날씨")["총계"].mean())

# 조건으로 filter
print(df[(df["청소년"] > 60) & (df["어린이"] != 0)])

# 특정 data 추출
print(df.loc[[1, 5, 12], ["날씨", "어린이"]])
print(df.loc[1:5, ["날짜", "어른", "청소년"]])

# 정렬
df_sorted = df.sort_values("총계", ascending=False)
print(df_sorted.head())

# 인덱스 재지정
df_sorted = df_sorted.reset_index(drop=True)
print(df_sorted.head())

# 정렬 후 iloc으로 접근
print(df_sorted.iloc[0]["날짜"])

# 컬럼 삭제
df = df.drop(["유료합계", "무료합계"], axis=1)

# 컬럼 이름 변경
df = df.rename(columns={"총계": "총입장객수"})

# 결측치 확인
print(df.isnull().sum())

# 결측치 채우기
df = df.fillna(int(df["청소년"].mean()))

# 결측치 삭제
# df = df.dropna(subset="청소년", ignore_index=True)

# map으로 값 변환
df["공휴일"] = df["공휴일"].map({"O": 1, "X": 0})
print(df["공휴일"].value_counts())


# merge: 특정 컬럼 기준으로 합치기
df_info = pd.DataFrame({
    "날짜": df["날짜"],
    "날씨": df["날씨"]
})
df_visitors = pd.DataFrame({
    "날짜": df["날짜"],
    "총입장객수": df["총입장객수"]
})
df_merged = pd.merge(df_info, df_visitors, on="날짜")
print(df_merged.head())
```
