```python
# transform : 데이터를 머신러닝 알고리즘 학습에 최적화된 형태로 변환시키는 것

# callable(호출 가능한 객체) : ()를 사용할 수 있는 모든 객체(()을 호출이라 생각하면 된다.)
# transform은 특징을 변형하고 target_transform은 정답을 변형한다.
# 모든 TorchVision의 데이터셋은 변형 로직을 갖는 매개변수를 두 개 가지는데 둘 다 호출 가능한 객체이다.

# one-hot(원-핫) 인코딩 : 데이터를 벡터화할 때 해당하는 데이터의 종류를 1 나머지는 0으로 표현하는 방식
# FashionMNIST의 특징과 정답을 제대로 학습시키기 위해선 특징은 정규화하고 정답은 원-핫으로 부호화된 텐서로 바꿔야 한다.
# 이러한 변형을 위해 ToTensor와 lambda를 사용한다.
```

```python
import torch
from torchvision import datasets
from torchvision.transforms import ToTensor, Lambda

# ToTensor : PIL이미지와 ndarray를 FloatTensor로 변환하여 이미지 픽셀을 0.0 ~ 1.0으로 조정(scale)한다.(정규화의 초석)
# FashionMNIST의 이미지는 PIL이미지 형식
ds = datasets.FashionMNIST(
  root="data",
  train=True,
  download=True,
  transform=ToTensor(),
  target_transform=Lambda(lambda y: torch.zeros(10, dtype=torch.float).scatter_(0, torch.tensor(y), value=1))
)
```
