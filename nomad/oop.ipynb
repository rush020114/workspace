```python
# 클래스 만들기
class Dog:
  # class바깥은 함수 클래스 안은 메서드이다.
  # 메서드 생성 문법에 __가 필요할 때도 있다.
  # 어떤 메서드이든 첫번째 매개변수로 자기자신을 참조하고 있다.
  # 그리고 이 첫번째 매개변수는 반드시 매개변수로 받아야 한다.
  # __init__ 메서드는 객체 생성 시 값을 초기화할 수 있다.
  def __init__(self, name, breed, age):
    self.name = name
    self.age = age
    self.breed = breed

  # 공통으로 가지는 메서드가 된다.
  def sleep(self):
    print("zzz...")

  def introduce(self):
    print(f"My name is {self.name} and I am a {self.breed}")

# 상속
# 상속 문법 - 클래스명(부모클래스)
class Puppy(Dog):
  # 상속으로 고정값을 커스터마이징할 수 있다.
  # super()키워드는 부모 클래스는 부모 클래스를 참조한다.
  # 부모 클래스의 생성자를 초기화할 수 있다.
  def __init__(self, name, breed):
    super().__init__(
      name, 
      breed, 
      0.1
    )
    # 공통 변수는 위의 방식으로 초기화하고
    # 개별 변수를 __init으로 초기화할 수 있다.
    self.spoiled = True
  
  # 객체의 참조값을 대체할 문자열을 return
  def __str__(self):
    return f"{self.breed} puppy named {self.name}"

  # __ 없는 메서드도 가능하다.
  # 물론 자기 자신을 참조하는 self매개변수는 항상 필요하다.
  def woof_woof(self):
    print("Woof Woof!")

  # class 안이어도 자기 자신으로 class의 메서드를 호출한다.
  def introduce(self):
    self.woof_woof()
    super().introduce()
    self.woof_woof()

  # pass : 계속 진행이라는 뜻
  # pass

class GuardDog(Dog):
  def __init__(self, name, breed):
    super().__init__(
      name, 
      breed, 
      5
    )
    self.aggressive = True

  def rrrrr(self):
    print("stay away!")

  def introduce(self):
    super().introduce()
    self.rrrrr()

""" 
  ruffus라는 객체 생성 
  객체 생성 문법 - 클래스명()
"""
ruffus = Puppy(
  name="ruffus", 
  breed="Beagle", 
)
bibi = GuardDog(
  name="Bibi", 
  breed="Dalmatian", 
)
ruffus.woof_woof()
bibi.rrrrr()

ruffus.sleep()
bibi.sleep()

ruffus.introduce()
bibi.introduce()
```

```python
class Player:
  def __init__(self, name, team):
    self.name = name
    self.xp = 1500
    self.team = team

  def introduce(self):
    print(f"Hello! I'm {self.name} and I play for {self.team}")

class Team:
  def __init__(self, team_name):
    self.name = team_name
    self.players = []

  # player이름만 넣으면 팀명이 자동으로 들어가는 구조이다.
  def add_player(self, name):
    new_player = Player(name, self.name)
    self.players.append(new_player)

  def show_player(self):
    for player in self.players:
      player.introduce()

  def delete_player(self, name):
    del self.players[0]
        
  

team_x = Team("Team X")

team_blue = Team("Team Blue")

team_x.add_player("Nico")

team_blue.add_player("Lynn")


team_x.delete_player("Nico")
team_x.show_player()
team_blue.show_player()
```
