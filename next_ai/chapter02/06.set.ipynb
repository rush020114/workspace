# set

```python
# 집합
# 집합은 순서가 없고 중복을 제거한다
set1 = {2, 3, 4}
set2 = set([2, 3, 4])
set3 = {2, 3, 2, 4}

num_set = {1, 3, 5, 7}
num_set.add(9)
# 여러 개의 원소를 추가(인자로 list, set 다 가능)
num_set.update([11, 13, 15])
# 원소가 없으면 오류
num_set.remove(15)
# 원소가 없어도 오류 안 남
num_set.discard(17)

# in, len 사용 가능
print(1 in num_set)
print(len(num_set))
```

# 집합 연산

```python
set1 = {1, 3, 5, 7}
set2 = {3, 5, 7, 9}
# 합집합
union = set1 | set2
# 교집합
intersection = set1 & set2
# 차집합
diff = set1 - set2
# xor
xor = set1 ^ set2

print(union, intersection, diff, xor)
```
