```python
# /는 실수 나눗셈, //은 몫 나눗셈, **은 제곱
print(5 // 2, 2 ** 4)
# 문자열 덧셈은 합해지고 곱셈은 반복이다.
print("안녕" + "하세요")
print("안녕" * 3)
```

```python
# 문자열도 리스트처럼 여러 원소로 이루어져 있어 []로 접근 가능하다.
alpha = "Ready"

# indexing: index를 이용해서 문자열과 리스트의 특정 문자열을 가져오는 방법
print(alpha[1])

# slicing: 문자열과 리스트의 원소 '일부'를 들고오는 것
# 문법: String/list[시작 인덱스:종료 인덱스] - 이상 미만의 법칙
print(alpha[2:4])
```

```python
# append: 오직 한 개의 자료를 마지막에 넣을 수 있는 함수
a = []
b = ['a', 'b', 'c']

a.append(10)
b.append('d')
print(a, b)

# insert: 몇 번째 인덱스에 넣고 싶은 자료를 하나만 넣을 수 있는 함수
# 첫 번째 매개변수: 넣을 인덱스, 두 번째 매개변수: 들어갈 값
c = [1, 2, 4, 5]
c.insert(2, 3)
print(c)

# 원소 하나를 지울 수 있는데 중복될 경우 첫 번째에 발견된 원소가 지워진다.
my_list = [3, 1, 2, 3]
my_list.remove(3)
print(my_list)

# 정렬 함수
# 숫자, 문자가 함께 있으면 사용 불가
e = [2, 1, 4, 3]
f = ['d', 'c', 'b']

e.sort()
f.sort()
print(e, f)

# 꺼내기 함수 
my_list = [2, 3, 5, 7, 11]
# 0번째 요소 꺼내기
print(my_list.pop(0))
# 매개변수가 없으면 0번째 요소를 빼냄
print(my_list.pop())
print(my_list)

# 함수 안에 있는 자료가 몇 개인지 알 수 있는 함수
my_seq = [2, 2, 2, 4, 4]
print(my_seq.count(4))
my_str = "elice and alice"
print(my_str.count('e'))

# 문자열을 쪼개 리스트화하는 함수
# 매개변수 미전달 시 공백을 기준으로 처리
my_str = "1 2 3 4 5"
print(my_str.split())
element = "Na, Mg, Al, Si"
print(element.split(','))

# 리스트를 합쳐서 문자열을 만드는 함수
my_list = ['e', 'l', 'i', 'c', 'e']
print("".join(my_list))
station = ['서울', '대전', '대구']
print("->".join(station))
```
