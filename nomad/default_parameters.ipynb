```python
# 기본 매개변수를 주어 오류가 나는 것을 방지할 수 있다.
def say_hello(my_name = "anonymous"):
  print("My name is", my_name)
say_hello()
```

```python
def plus(a = 0, b = 0):
  print(a + b)

def minus(a = 0, b = 0):
  print(a - b)

def multiply(a = 0, b = 0):
  print(a * b)

def divide(a = 0, b = 1):
  print(a / b)

def square(a = 0, b = 0):
  print(a ** b)

plus()
minus()
multiply()
divide()
square()
```
