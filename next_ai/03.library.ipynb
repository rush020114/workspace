```python
# 파이썬의 가상환경: 프로젝트마다 패키지를 독립적으로 설치할 수 있게 해주는 공간
# npm 같은 경우는 폴더마다 독립적으로 설치 가능하지만 python은 컴퓨터 자체에 설치돼서 모든 프로젝트와 공유
# 가상환경을 사용하면 프로젝트마다 독립적으로 다른 버전의 패키지를 설치할 수 있음

# 의존성: 특정 패키지가 동작하기 위해 필요한 다른 패키지
# pip 패키지 관리자는 의존성을 알아서 판단해주고 같이 설치해준다.

# 가상환경은 가상환경은 생성 후 활성화해주어야 한다.
# venv 가상환경은 생성 시 현재 버전의 파이썬으로 생성(다른 버전이 필요하다면 conda 사용을 추천)
# 만약 그만 쓰고 싶다면 비활성화 또는 삭제를 해준다.

# 작업: Windows | Linux/Mac
# 생성: python -m venv 이름 | python -m venv 이름
# 활성화: 이름\Scripts\activate | source 이름/bin/activate
# 비활성화: deactivate | deactivate
# 삭제: rmdir /s 이름 | rm -rf 이름

# 패키지를 한 번에 설치하고 싶으면
# requirements.txt(암묵적 이름)에 특정 형식으로 패키지들을 입력하고
# pip install -r requirements.txt
# 내 패키지 목록을 전달하고 싶다면(txt 형식으로 생성됨)
# pip freeze > requirements.txt
```

```python
# 파일 기반 라이브러리 설치 방법
# 패키지 설치 방법 4가지

# 1. pip (가장 일반적)
# 사전작업: 없음 / 인터넷 필요
# 내부적으로 PyPI에서 wheel 파일을 자동으로 다운로드 후 설치
# pip install requests

# 2. wheel
# 사전작업: PyPI에서 .whl 파일 미리 다운로드 / 인터넷 불필요
# 미리 컴파일된 파일이라 바로 설치 가능
# pip install requests.whl

# 3. tar.gz
# 사전작업: PyPI에서 .tar.gz 파일 미리 다운로드 / 인터넷 불필요
# 소스코드라 컴파일 필요, wheel이 없을 때 사용
# tar -zxvf requests.tar.gz → cd requests → pip install .

# 4. git clone
# 사전작업: 없음 / 인터넷 필요
# PyPI에 없는 패키지일 때 github에서 소스코드 받아서 설치
# git clone 주소 → cd requests → pip install .

# 우선순위
# 인터넷 되면    → pip
# 인터넷 안 되면 → wheel 먼저
# wheel 없으면  → tar.gz 또는 git clone
```
